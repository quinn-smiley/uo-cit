```python
# Quinn Smiley
# CIT 383
# May 4th, 2026

# *Started with my own code and had Cursor refine it

import random

movies_by_genre = {
    "Horror": ["The Conjuring", "Sinister", "Evil Dead Rise"],
    "Romantic Comedy": ["10 Things I Hate About You", "The Devil Wears Prada", "Practical Magic"],
    "Action/Adventure": ["Mission: Impossible", "Iron Man", "Ocean's 11"],
    "Mystery": ["Clue", "Knives Out", "Rear Window"],
    "Feel-good": ["Legally Blonde", "Happy Gilmore", "Ferris Bueller's Day Off"],
    "Comedy": ["Superbad", "The Blues Brothers", "The Hangover"],
    "Thriller": ["Se7en", "The Silence of the Lambs", "Shutter Island"],
    "Family-friendly": ["Shrek", "Cars", "The Incredibles"],
}

genres = list(movies_by_genre.keys())

# remembers what we've already suggested this run
suggested = {genre: set() for genre in genres}

def display_genres():
    for i, g in enumerate(genres, start=1):
        print(f"{i}. {g}")

def pick_genre(user_choice: str):
    s = user_choice.strip().lower()

    if s.isdigit():
        n = int(s)
        if 1 <= n <= len(genres):
            return genres[n - 1]
        return None

    for genre in genres:
        if s in genre.lower():
            return genre

    return None

def suggest_movie(genre: str):
    remaining = [m for m in movies_by_genre[genre] if m not in suggested[genre]]
    if not remaining:
        return None
    movie = random.choice(remaining)
    suggested[genre].add(movie)
    return movie

display_genres()

while True:
    inp = input("Pick a genre (number or name), or 'q' to quit: ").strip().lower()
    if inp == "q":
        break

    genre = pick_genre(inp)
    if not genre:
        print("Please pick a genre from the list.")
        continue

    movie = suggest_movie(genre)
    if not movie:
        print(f"You've seen all my {genre} suggestions!")
        continue

    while True:
        print(f"Suggestion ({genre}): {movie}")
        ans = input("Seen it? (y/n) ").strip().lower()

        if ans == "n":
            break
        if ans == "y":
            movie = suggest_movie(genre)
            if not movie:
                print(f"You're out of {genre} movies in my list.")
                break
        else:
            print("Type 'y' or 'n'.")
```
